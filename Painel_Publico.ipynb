# Cadastrar.IA | Painel público de BI e Analytics
Visualize os resultados documentais agregados e o estudo complementar de varejo. Mantenha `docs/index.html` no diretório de execução ou forneça esse arquivo público ao ambiente com o nome `index.html`.

```python
from html import escape
from pathlib import Path
from IPython.display import HTML, display

opcoes = (Path('docs/index.html'), Path('index.html'))
arquivo = next((caminho for caminho in opcoes if caminho.is_file()), None)
if arquivo is None:
    raise FileNotFoundError('Forneça o index.html público gerado em docs.')

conteudo = escape(arquivo.read_text(encoding='utf-8'), quote=True)
display(HTML(
    '<iframe title="Painel público do Cadastrar.IA" sandbox="allow-scripts" '
    'style="width:100%;height:900px;border:0" '
    f'srcdoc="{conteudo}"></iframe>'
))
```
